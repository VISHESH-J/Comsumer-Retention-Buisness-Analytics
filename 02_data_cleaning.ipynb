```python
import pandas as pd
import os

RAW_PATH = "../data/csv files"
CLEAN_PATH = "../data/cleaned"

os.makedirs(CLEAN_PATH, exist_ok=True)
```

```python
orders = pd.read_csv(
    f"{RAW_PATH}/olist_orders_dataset.csv"
)

orders_clean = orders.copy()
```

```python
date_columns = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]

for col in date_columns:
    orders_clean[col] = pd.to_datetime(
        orders_clean[col],
        errors='coerce'
    )
```

```python
orders_clean['order_id'] = orders_clean['order_id'].str.strip()
orders_clean['customer_id'] = orders_clean['customer_id'].str.strip()
orders_clean['order_status'] = orders_clean['order_status'].str.strip().str.lower()
```

```python
orders_clean.to_csv(
    f"{CLEAN_PATH}/olist_orders_clean.csv",
    index=False
)
```

```python
customers = pd.read_csv(
    f"{RAW_PATH}/olist_customers_dataset.csv"
)

customers_clean = customers.copy()
```

```python
customer_text_columns = [
    'customer_id',
    'customer_unique_id',
    'customer_zip_code_prefix',
    'customer_city',
    'customer_state'
]

for col in customer_text_columns:
    customers_clean[col] = customers_clean[col].astype('string').str.strip()
```

```python
customers_clean.to_csv(
    f"{CLEAN_PATH}/olist_customers_clean.csv",
    index=False
)
```
