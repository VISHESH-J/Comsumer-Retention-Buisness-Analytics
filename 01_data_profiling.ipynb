```python
orders_clean = orders.copy()
```

```python
date_columns = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]

for col in date_columns:
    orders_clean[col] = pd.to_datetime(
        orders_clean[col],
        errors='coerce'
    )
```

```python
orders_clean[date_columns].isnull().sum()
```

```python
orders_clean.dtypes
```

```python
invalid_approval = (
    orders_clean['order_approved_at'].notna() &
    (orders_clean['order_approved_at'] < orders_clean['order_purchase_timestamp'])
)

invalid_approval.sum()
```

```python
invalid_carrier = (
    orders_clean['order_delivered_carrier_date'].notna() &
    (orders_clean['order_delivered_carrier_date'] < orders_clean['order_purchase_timestamp'])
)

invalid_carrier.sum()
```

```python
invalid_customer_delivery = (
    orders_clean['order_delivered_customer_date'].notna() &
    (orders_clean['order_delivered_customer_date'] < orders_clean['order_purchase_timestamp'])
)

invalid_customer_delivery.sum()
```

```python
invalid_delivery_sequence = (
    orders_clean['order_delivered_customer_date'].notna() &
    orders_clean['order_delivered_carrier_date'].notna() &
    (orders_clean['order_delivered_customer_date'] <
     orders_clean['order_delivered_carrier_date'])
)

invalid_delivery_sequence.sum()
```

```python
orders_clean.groupby('order_status')[
    [
        'order_approved_at',
        'order_delivered_carrier_date',
        'order_delivered_customer_date'
    ]
].count()
```

```python
orders_clean['order_status'].value_counts()
```

```python
orders_clean['invalid_date_sequence'] = (
    invalid_approval |
    invalid_carrier |
    invalid_customer_delivery |
    invalid_delivery_sequence
)
```

```python
orders_clean['invalid_date_sequence'].value_counts()
```
